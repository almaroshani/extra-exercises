```python
df19 = df19.rename(columns={"Country or region": "Country"})

h15 = df15[["Country", "Happiness Score"]]
h19 = df19[["Country", "Score"]]

inner = pd.merge(h15, h19, how="inner", on="Country")
print(len(h15), len(h19), len(inner))

inner["change"] = inner["Score"] - inner["Happiness Score"]
inner.sort_values("change", ascending=False).head()
```

```python
left = h15.merge(h19, how="left", on="Country")
left[left["Score"].isna()]

right = h15.merge(h19, how="right", on="Country")
right[right["Happiness Score"].isna()]

#length of left is equal to h15!
print(len(left), len(h15))
print(len(right), len(h19))
```
